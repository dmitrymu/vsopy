# Simplified Transform Analysis
## Simulation Environment

Simplified transform method which claculates transformation coefficients from 
a couple of light frames in different bands explained in 
(Gary B.L., CCD Transformation Equations for Use with Single-Image Photometry)[https://researchers.usask.ca/gordon-sarty/documents/astronomy/transf.pdf].  This method requires relatively big number of 
standard stars in the frame.

To calculate transformed magnitude of the target star from frames in bands 
A and B given a comparison star and a number of standard stars we calculate 
slopes of three linear regressions:

$$\begin{aligned}
a - b = T_{ab} (A - B) + Z_{ab} \\
A - a = T_a (A - B) + Z_a       \\
B - b = T_b (A - B) + Z_b
\end{aligned}$$

where $A$ and $B$ are standard magnitudes, $a$ and $b$ are instrumental
magnitudes.

Let's create a test sequence of standard stars for bands A and B first.

```python
from vsopy.util import ValErr
from experiment import Experiment

SEED = 42

EXPERIMENT = Experiment(SEED)

SEQUENCE = EXPERIMENT.generate_sequence(
    Experiment.GenParams(
        num_stars=20,
        mag=(10.0, 15.0),
        ci=(-0.5, 1.0),
        uncert=ValErr(0.05, 0.05)))
SEQUENCE.stars
```

Select the brightest star in the band A in the sequence to be used as the comparison star

```python
SEQUENCE.comp_star
```

The rest of sequence is the ensemble to be used for transformation

```python
SEQUENCE.ensemble
```

Now simulate masurement of instrumental magnitudes.  We assume that instrumental magnitudes 
in both bands are offset by the same bias value and that the measurement
adds random error (scatter).

```python
from measurement import SimpleMeasurement

MEASUREMENT = EXPERIMENT.create_measurement(
    SimpleMeasurement,
    params = SimpleMeasurement.Params(
        offset=1.0,
        scatter=0.1,
        uncert=ValErr(0.07, .07)))
MEASUREMENT(SEQUENCE)
```

Now let's simulate measurements of the sequence and calculate transforms

```python
EXPERIMENT.create_transforms(
    SEQUENCE,
    MEASUREMENT,
    num_measurements=5)
```
