# Simplified transform analysis
## Effect of Sequence Properties on Transform Stability

Given that transform values are slopes of linear regressions over
stars in the sequence, we should expect that the following sequence
properties are affecting transform stability:

1. Number of stars
1. Magnitude range
1. Color index range

In all cases bigger value of the propery should offer better stability.

```python
from vsopy.util import ValErr
from measurement import SimpleMeasurement
from paramgrid import SequenceParamGrid

SEED = 42

EXP1 = SequenceParamGrid(SEED,
                         SequenceParamGrid.GridParams(
                             num_stars=[5, 10, 15, 20],
                             mag_range=[2, 3, 4, 5,],
                             ci_range=[0.5, .7, 1.0, 1.2, 1.5]),
                         num_attempts=5,
                         num_measurements=100)
MEASUREMENT = EXP1.create_measurement(
    SimpleMeasurement,
    params=SimpleMeasurement.Params(
        offset=1.0,
        scatter=0.05,
        uncert=ValErr(0.03, 0.03)
    ))
```

```python

RESULTS = EXP1.run(MEASUREMENT)
RESULTS.to_pandas()
```

```python
RESULTS.to_pandas().melt(id_vars=['Nstars', 'Mag_range', 'CI_range'], var_name='transform', value_name='y')
```

```python
from plotnine import ggplot, aes, geom_point, facet_grid, theme

(ggplot(RESULTS.to_pandas().melt(id_vars=['Nstars', 'Mag_range', 'CI_range'], var_name='transform', value_name='y'))
    + aes(x='CI_range', y='y', color='Mag_range') + geom_point()
 + facet_grid('transform ~ Nstars', scales='free_y')
    #+ facet_wrap('~ transform + Nstars', scales='free_y')
#    + scale_x_log10()
    + theme(figure_size=(10, 10)))
```
